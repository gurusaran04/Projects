```python
x=df.drop('is_fraud',axis=1)
y=df['is_fraud']
```

```python
y
```

```python
import numpy as np
from sklearn.preprocessing import MinMaxScaler

# Combine x and y to drop rows with NaN consistently
data = pd.DataFrame(x) # Convert x back to DataFrame to use dropna
data['y'] = y.values # Add y as a column
data_cleaned = data.dropna()

x = data_cleaned.drop('y', axis=1).values # Get cleaned x values as numpy array
y = data_cleaned['y'].values # Get cleaned y values as numpy array

scaler = MinMaxScaler()
x = scaler.fit_transform(x)
x
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=47)
```

```python
from sklearn.naive_bayes import GaussianNB
r = GaussianNB()
r.fit(x_train,y_train)
y_pred=r.predict(x_test)
y_pred
```

```python
import sklearn.naive_bayes

print("Other modules/classes in sklearn.naive_bayes:")
for name in dir(sklearn.naive_bayes):
    if not name.startswith('_') and name[0].isupper(): # Filter for public classes/modules
        print(name)
```

```python
from sklearn.metrics  import confusion_matrix,accuracy_score,classification_report
y_pred_binary = (y_pred > 0.5).astype(int)
print(confusion_matrix(y_test,y_pred_binary))
print(accuracy_score(y_test,y_pred_binary))
print(classification_report(y_test,y_pred_binary))
```
